# 1. Numpy (Sort)

```python
import numpy as np
```

```python
np.random.seed(1)    # .seed use for constent  
a = np.random.randint(1,100,8)
print("original a:",a)
```

```python
print(np.sort(a))  # .sort use for shorting
```

```python
print(a)
```

```python
print(np.sort(a)[::-1])
```

```python
b = np.random.randint(1,100,12).reshape(3,4)
print(b)
print(np.sort(b,axis=0)) # row
```

# 2. np.append

```python
a = np.array([10,20,30])
print(a)
print(np.append(a,200))
```

```python
np.random.seed(2)
b = np.arange(8).reshape(4,2)
print(b)
```

```python
print(b.shape)
extra_col = np.random.randint(1,10,(b.shape[0],1))
print(b)
print(extra_col)
print(np.append(b,extra_col,axis=1)) # column
```

```python
new_row = np.array([[99,99]])
print(new_row)
print(np.append(b,new_row,axis=0))
```

## 3 np.Concatenate

```python
c = np.arange(6).reshape(2,3)
d= np.arange(6,12).reshape(2,3)
print(c)
print(d)
```

```python
print(np.concatenate((c,d),axis=0)) # row
```

```python
print(np.concatenate((c,d),axis=1)) # axis 1 column
```

```python
e1 =np.array([1,2,3])
e2 =np.array([4,5])
print(np.concatenate((e1,e2)))
```

# 4 np.unique

```python
arr = np.array([1,1,2,2,3,3,4,4,4,5])
print(arr)
```

```python
print(np.unique(arr))
```

```python
print(arr)
```

```python
values ,count = np.unique(arr,return_counts=True)
print(values)
print(count)
```

```python
values,idx = np.unique(arr,return_index=True)
print(values)
print(idx)
```

# 5. np.expand_dims

```python
a = np.array([1,2,3,4])
print(a)
print(a.shape)
print(np.expand_dims(a,axis=0).shape)
```

```python
expanded = np.expand_dims(a,axis=1)
print(expanded)
```

# 6. np.where

```python
np.random.seed(3)
a = np.random.randint(1,100,10)
print(a)
print(np.where(a>50))
```

```python
print(np.where(a>50,0,a))
```

```python
print(np.where(a%2==0,0,a))
```

# 7. np.argmax and np.argmin

```python
np.random.seed(3)
a = np.random.randint(1,100,10)
print(a)
print(np.argmax(a))
print(np.argmin(a))
```

```python
b = np.random.randint(1,100,12).reshape(3,4)
print(b)
print(np.argmax(b,axis=0)) # column
```

```python
print(np.argmax(b,axis=1))
```
