```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
cafe = pd.read_csv('chai_cafe_sales.csv')
marks = pd.read_csv('student_marks.csv')
city = pd.read_csv('city_monthly_temperature.csv',index_col='city')
```

# 1 ) Axis Level Vs Figure level function:

## Axis Level Function 

```python
marks.head()
```

```python
ax = sns.histplot(data=marks,x='maths')
print(type(ax))
```

# Figure Level Function

```python
g = sns.displot(data=marks,x='maths',kind='hist')
print(type(g))
```

```python
fig,axes = plt.subplots(1,2,figsize=(10,4))
sns.histplot(data=marks, x='maths',ax=axes[0])
sns.scatterplot(data=marks,x='study_hours',y='maths',ax= axes[1])
```

# Relational Plot : 2) Scatter Plot : sxatterplot() vs relplot(kind='scatter')

```python
sns.scatterplot(data=cafe , x='temperature_c',y='cold_coffee_cups')
```

```python
sns.scatterplot(data=cafe,x='temperature_c',y='cold_coffee_cups',hue='branch')
```

```python
plt.figure(figsize=(7,4.5))
sns.scatterplot(data=cafe,x='temperature_c',y='cold_coffee_cups',hue='branch',style='branch',size='total_sales_rs')
plt.title('Hot Days Sells More cold coffee')
```

# Figure Level Plot : replot (kind= 'scatter')

```python
sns.relplot(data=cafe,x='temperature_c',y='cold_coffee_cups',hue='branch',kind='scatter')
```

```python
g= sns.relplot(data=cafe,x='temperature_c',y='cold_coffee_cups',hue='branch',kind='scatter',col='branch',
              height=4,aspect=1.1) # width = hight * aspect
g.set_axis_labels('temperature(dog c','cold_coffee_cups')
```

# 3) Line Plot : lineplot() vs relplot(kind='line')

```python
sns.lineplot(data=cafe,x='day',y='total_sales_rs',hue='branch')
```

```python
plt.figure(figsize=(8,4))
sns.lineplot(data=cafe,x='day',y='total_sales_rs',hue='branch',marker='o')
plt.title('Daily Sales')
plt.ylabel('Total sales (Rs)')
```

```python
plt.figure(figsize=(8,4))
sns.lineplot(data=cafe,x='weekday',y='total_sales_rs',hue='branch',marker='o')
plt.title('Average Sales by weekday(2 Days)')
```

```python
plt.figure(figsize=(8,4))
sns.lineplot(data=cafe,x='weekday',y='total_sales_rs',hue='branch',marker='o',errorbar=None)
plt.title('Average  Sales By Weekday')
```

```python
g=sns.relplot(data=cafe,x='day',y='total_sales_rs',hue='branch',kind='line',marker='o',
             col='branch',height=3.8,aspect=1.2)
g.set_axis_labels('Day','Total sales (Rs')
```

# 4) Histogram : histplot() Vs displot(kind='hist')

```python
marks.head()
```

```python
sns.histplot(data=marks,x='maths')
```

```python
sns.histplot(data=marks,x='maths',bins=8,color='teal')
```

```python
sns.histplot(data=marks,x='maths',bins=8,color='teal',kde=True)   # kernal 
plt.title('maths marks of 40 student')
```

```python
sns.histplot(data=marks,x='maths',bins=8,hue='class',multiple='stack')
plt.title('maths marks -- 10A Vs 10B')
```

```python
sns.histplot(data=marks,x='maths',bins=8,hue='class',multiple='dodge')
plt.title('maths marks -- 10A Vs 10B')
```

# Figure Level histogram : displot(kind='hist')

```python
g=sns.displot(data=marks,x='maths',kind='hist',bins=8,col='class',hue='gender',multiple='stack',height=3.8,aspect=1.1)
g.set_axis_labels('maths marks','number of student')
```

 # 5)KDE Plot: kernal density estimate: ldeplot() vs displot(kind='kde') 

```python
sns.kdeplot(data=marks,x='maths')
```

```python
sns.kdeplot(data=marks,x='maths',hue='class')
```

```python
sns.kdeplot(data=marks,x='maths',hue='class',fill=True,alpha=0.4,linewidth=4)
plt.title('maths marks --10A vs 10B')
```

# Figure Level function : displot(kind='kde')

```python
g = sns.displot(data=marks,x='science',kind='kde',hue='gender',fill=True,col='class',height=3.8,aspect=1.1)
g.set_axis_labels('Science marks','Density')
```

# Rugplot : rugplot() vs displot(rug=True)

```python
plt.figure(figsize=(7,2))
sns.rugplot(data=marks,x='study_hours',height=0.2)
```

```python
marks.head()
```

```python
sns.kdeplot(data=marks,x='study_hours',fill=True)
sns.rugplot(data=marks,x='study_hours',height=0.08)
plt.title('Dailt study hours of 40 Student')
```

```python
sns.kdeplot(data=marks,x='study_hours',hue='class',fill=True)
sns.rugplot(data=marks,x='study_hours',hue='class',height=0.08)
plt.title('Study hours -- 10A vs 10B')
```
