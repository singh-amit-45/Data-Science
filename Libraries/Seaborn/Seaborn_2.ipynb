```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')

food = pd.read_csv('food_orders.csv')
gym = pd.read_csv('gym_members.csv')
```

```python
food.head()
```

# Axes Level Function :

```python
ax = sns.boxplot(data=food,x='day_type',y='delivery_min')
print(ax)
```

# Figure Level Function : catplot()

```python
g = sns.catplot(data=food,x='day_type',y='delivery_min',kind='box',col='food_type')
print(type(g))
```

# Categorial plots :

- a) Bar Plot : barplot() vs catplot(kind='bar')

```python
sns.barplot(data=food,x='cuisine',y='bill_rs')
```

```python
plt.figure(figsize=(8,4))
sns.barplot(data=food, x='cuisine', y='bill_rs',hue='food_type', errorbar=None)
plt.title('Average bill Veg Vs NonVeg')
```

```python
plt.figure(figsize=(8,4))
sns.barplot(data=food, x='cuisine', y='bill_rs',estimator='sum', errorbar=None,color='salmon')
g.set_axis_labels('cusion','Average Bill (rs)')
```

```python
g = sns.catplot(data=food , x='cuisine',y='bill_rs',kind='bar',hue='food_type',col='day_type',errorbar=None,height=3.8,aspect=1.2)
g.set_axis_labels('Cuisine','Average bill (rs)')
```

# b) Countplot - Countplot() vs Catplot(kind='count')

```python
sns.countplot(data = food, x='cuisine')
```

```python
top = food['cuisine'].value_counts().index
print(top)
```

```python
sns.countplot(data=food,x='cuisine',order=top,color='orange')
plt.title('Number of orders by Cuisine')
```

```python
top = food['cuisine'].value_counts().index
print(top)
```

```python
sns.countplot(data=food,y='cuisine',order=top,color='orange')
plt.title('  Number of Orders by cuisine')
```

```python
plt.figure(figsize=(7.5,4))
sns.countplot(data=food,x='day',hue='day_type',order=['mon','tue','wed','thu','fri','sat','sun'])
plt.title('Order per day of the week')
```

# Figure Level function : Catplot(kind='count')

```python
g = sns.catplot(data=food,x='cuisine',kind='count',col='food_type',color='orange',height=3.8,aspect=1.2)
g.set_axis_labels('Cuisine','Number of orders')
```

# c )Box plot: --boxplot() vs catplot(kind='box')

```python
sns.boxplot(data=food,x='day_type',y='delivery_min')
```

```python
plt.figure(figsize=(8,4))
sns.boxplot(data=food,x='cuisine',y='bill_rs',hue='cuisine',palette='Set2')
plt.title('Bill Amount spread by Cuisine')
```

```python
plt.figure(figsize=(8,4))
sns.boxplot(data=food,x='day_type',y='delivery_min',hue='food_type',width=0.6)
plt.title('Delivey time -- Weekday vs Weekend ')
```

# Figure Level boxplot : catplot(kind='box')

```python
g = sns.catplot(data=food,x='cuisine',y='delivery_min',kind='box',col='day_type',color='skyblue',height=3.8,aspect=1.25)
g.set_axis_labels('Cuisine', 'Delivery_time(min)')
```

```python
g = sns.catplot(data=food,x='cuisine',y='delivery_min',kind='bar',col='day_type',color='skyblue',height=3.8,aspect=1.25)
g.set_axis_labels('Cuisine', 'Delivery_time(min)')
```
