# 1. Numpy (Sort)

```python
import numpy as np
```

```python
np.random.seed(1)    # .seed use for constent  
a = np.random.randint(1,100,8)
print("original a:",a)
```

# 2. np.append

```python
a = np.array([10,20,30])
print(a)
print(np.append(a,200))
```

```python
np.random.seed(2)
b = np.arange(8).reshape(4,2)
print(b)
```

```python
print(b.shape)
extra_col = np.random.randint(1,10,(b.shape[0],1))
print(b)
print(extra_col)
print(np.append(b,extra_col,axis=1)) # column
```

```python
new_row = np.array([[99,99]])
print(new_row)
print(np.append(b,new_row,axis=0))
```

## 3 np.Concatenate

```python
c = np.arange(6).reshape(2,3)
d= np.arange(6,12).reshape(2,3)
print(c)
print(d)
```

```python
print(np.concatenate((c,d),axis=0)) # row
```

```python
print(np.concatenate((c,d),axis=1)) # axis 1 column
```

```python
e1 =np.array([1,2,3])
e2 =np.array([4,5])
print(np.concatenate((e1,e2)))
```

# 10 np.histogram

```python
np.random.seed(3)
a = np.random.randint(1,100,10)
print(a)
```

```python
counts,edges = np.histogram(a,bins = [0,50,100])
print(counts)
print(edges)
```

```python
counts,edges = np.histogram(a,bins=[1,25,50,75,100])
print(counts)
print(edges)
```

```python
counts,edges = np.histogram(a,bins = 4)
print(counts)
print(edges)
print(a)
```

# 12 np.isin

```python
a = np.array([15,20,35,40,55,60,75,80])
items = [20,40,60,80,100]
print(a)
print(np.isin(a,items))
```

```python
print(a[np.isin(a,items)])
```

```python
print(a[~np.isin(a,items)])
```

```python
b = np.arange(12).reshape(3,4)
print(b)
```

```python
print(np.isin(b,[0,5,10]))
```

# 15 Set Function

```python
m = np.array([1,2,3,4,5])
n =np.array([3,4,5,6,7])
print(m)
print(n)
```

```python
print(np.union1d(m,n)) # delete the duplicate 
```

```python
print(np.intersect1d(m,n))
```

```python
print(np.setdiff1d(m,n))
```

```python
print(np.setxor1d(m,n))
```

```python
print(m)
print(n)
```
