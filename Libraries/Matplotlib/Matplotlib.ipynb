```python
import matplotlib.pyplot as plt 
import pandas as pd 
import numpy as np 
```

```python
steps = pd.read_csv('daily_steps_tracker.csv')
```

```python
steps.head()
```

## 1 ) Line plot : plt.plot()

```python
plt.plot(steps['day'],steps['rahul_steps'])
plt.show()
```

```python
plt.plot(steps['day'],steps['rahul_steps'])
plt.plot(steps['day'],steps['sneha_steps'])
plt.show()
```

```python
plt.plot(steps['day'],steps['rahul_steps'])
plt.plot(steps['day'],steps['sneha_steps'])
plt.title('daily Steps -- Rahul vs Sneha(2 Weeks)')
plt.xlabel('Day')
plt.ylabel('steps walked')
plt.show()
```

```python
plt.plot(steps['day'],steps['rahul_steps'],color = '#e8590C',linestyle='solid',linewidth=2.5)
plt.plot(steps['day'],steps['sneha_steps'],color = '#1971C2',linestyle='dashed',linewidth=2)
plt.title('daily Steps -- Rahul vs Sneha(2 weeks)')
plt.xlabel('Day')
plt.ylabel('steps walked')
plt.show()
```

```python
plt.plot(steps['day'],steps['rahul_steps'],color = '#e8590C',linestyle='solid',linewidth=2.5,marker = 'o',markersize =7)
plt.plot(steps['day'],steps['sneha_steps'],color = '#1971C2',linestyle='dashed',linewidth=2,marker ='s',markersize=6)
plt.title('daily Steps -- Rahul vs Sneha(2 weeks)')
plt.xlabel('Day')
plt.ylabel('steps walked')
plt.show()
```

```python
plt.plot(steps['day'],steps['rahul_steps'],color = '#e8590C',linestyle='solid',linewidth=2.5,marker = 'o',markersize =7,label = 'Rahul')
plt.plot(steps['day'],steps['sneha_steps'],color = '#1971C2',linestyle='dashed',linewidth=2,marker ='s',markersize=6,label = 'Sneha')
plt.title('daily Steps -- Rahul vs Sneha(2 weeks)')
plt.xlabel('Day')
plt.ylabel('steps walked')
plt.legend(loc='upper left')
plt.show()
```

```python
plt.figure(figsize=(9,4.5))
plt.plot(steps['day'],steps['rahul_steps'],color = '#e8590C',linestyle='solid',linewidth=2.5,marker = 'o',markersize =7,label = 'Rahul')
plt.plot(steps['day'],steps['sneha_steps'],color = '#1971C2',linestyle='dashed',linewidth=2,marker ='s',markersize=6,label = 'Sneha')
plt.title('daily Steps -- Rahul vs Sneha(2 weeks)')
plt.xlabel('Day')
plt.ylabel('steps walked')
plt.legend(loc='upper left')
plt.xticks(steps['day'],steps['weekday'])
plt.show()
```

```python
plt.figure(figsize=(9,4.5))
plt.plot(steps['day'],steps['rahul_steps'],color = '#e8590C',linestyle='solid',linewidth=2.5,marker = 'o',markersize =7,label = 'Rahul')
plt.plot(steps['day'],steps['sneha_steps'],color = '#1971C2',linestyle='dashed',linewidth=2,marker ='s',markersize=6,label = 'Sneha')
plt.axhline(10000,color = 'green',linestyle = ':', linewidth = 5, label ='10000 Goal')
plt.title('daily Steps -- Rahul vs Sneha(2 weeks)')
plt.xlabel('Day')
plt.ylabel('steps walked')
plt.legend(loc='upper left')
plt.xticks(steps['day'],steps['weekday'])
plt.grid(alpha = 1)
plt.show()
```

```python
plt.figure(figsize=(9,4.5))
plt.plot(steps['day'],steps['rahul_steps'],color = '#e8590C',linestyle='solid',linewidth=2.5,marker = 'o',markersize =7,label = 'Rahul')
plt.plot(steps['day'],steps['sneha_steps'],color = '#1971C2',linestyle='dashed',linewidth=2,marker ='s',markersize=6,label = 'Sneha')
plt.axhline(10000,color = 'green',linestyle = ':', linewidth = 2, label ='10000 Goal')
plt.annotate('sunday trek',xy=(14,14100),xytext=(10,15000),arrowprops=dict(arrowstyle ='->',color='black'))
plt.title('daily Steps -- Rahul vs Sneha(2 weeks)')
plt.xlabel('Day')
plt.ylabel('steps walked')
plt.legend(loc='upper left')
plt.xticks(steps['day'],steps['weekday'])
plt.grid(alpha = 0.8)
plt.ylim(4000,16500)
plt.show()
```

# 2) Scatter plot :
- plt.scatter()

```python
food = pd.read_csv('food_delivery_orders.csv')
food.head()
```

```python
plt.scatter(food['distance_km'],food['delivery_time_min'])
plt.show()
```

```python
plt.scatter(food['distance_km'],food['delivery_time_min'])
plt.title('Distance Vs Delivery Time (120 orders)')
plt.ylabel('Distance Time (minutes)')
plt.xlabel('Distance (km)')
plt.show()
```

```python
plt.scatter(food['distance_km'],food['delivery_time_min'],color = '#7048E8',marker='o',alpha=0.6,edgecolors='red')
plt.title('Distance Vs Delivery Time (120 orders)')
plt.ylabel('Distance Time (minutes)')
plt.xlabel('Distance (km)')
plt.show()
```

```python
plt.scatter(food['distance_km'],food['delivery_time_min'],color = '#7048E8',marker='o',alpha=0.6,edgecolors='red',s=food['bill_amount']/3)
plt.title('Distance Vs Delivery Time (120 orders)')
# plt.colorbar(label ='Customer Rating')
plt.ylabel('Distance Time (minutes)')
plt.xlabel('Distance (km)')
plt.show()
```

```python
plt.scatter(food['distance_km'],food['delivery_time_min'],color = '#7048E8',marker='o',alpha=0.6,edgecolors='red',s=food['bill_amount']/3)
plt.title('Buble Size = Bill Amount')
# plt.colorbar(label ='Customer Rating')
plt.ylabel('Distance Time (minutes)')
plt.xlabel('Distance (km)')
plt.show()
```

```python
plt.scatter(food['distance_km'],food['delivery_time_min'],c= food['rating'],cmap='RdYlGn',alpha=0.6,edgecolors='black',s=60)
plt.title('Late order Get Bad Rating')
plt.colorbar(label ='Customer Rating')
plt.ylabel('Distance Time (minutes)')
plt.xlabel('Distance (km)')
plt.show()
```

```python
veg = food[food['food_type'] == 'Veg']
nonveg = food[food['food_type'] =='Non-Veg']
plt.scatter(veg['distance_km'],veg['delivery_time_min'],color = '#2F9E44',alpha=0.6,label='Veg')
plt.scatter(nonveg['distance_km'],nonveg['delivery_time_min'],color = '#E03131',alpha=0.6,label='NonVeg')
plt.title('Veg Vs Nonveg orders')
plt.xlabel('Distance_km')
plt.ylabel(('Delivery Time in minuts'))
plt.legend()
plt.show()
```
