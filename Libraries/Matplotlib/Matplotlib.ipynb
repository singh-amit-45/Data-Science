# 2) Scatter plot :
- plt.scatter()

```python
food = pd.read_csv('food_delivery_orders.csv')
food.head()
```

```python
plt.scatter(food['distance_km'],food['delivery_time_min'])
plt.show()
```

```python
plt.scatter(food['distance_km'],food['delivery_time_min'])
plt.title('Distance Vs Delivery Time (120 orders)')
plt.ylabel('Distance Time (minutes)')
plt.xlabel('Distance (km)')
plt.show()
```

```python
plt.scatter(food['distance_km'],food['delivery_time_min'],color = '#7048E8',marker='o',alpha=0.6,edgecolors='red')
plt.title('Distance Vs Delivery Time (120 orders)')
plt.ylabel('Distance Time (minutes)')
plt.xlabel('Distance (km)')
plt.show()
```

```python
plt.scatter(food['distance_km'],food['delivery_time_min'],color = '#7048E8',marker='o',alpha=0.6,edgecolors='red',s=food['bill_amount']/3)
plt.title('Distance Vs Delivery Time (120 orders)')
# plt.colorbar(label ='Customer Rating')
plt.ylabel('Distance Time (minutes)')
plt.xlabel('Distance (km)')
plt.show()
```

```python
plt.scatter(food['distance_km'],food['delivery_time_min'],color = '#7048E8',marker='o',alpha=0.6,edgecolors='red',s=food['bill_amount']/3)
plt.title('Buble Size = Bill Amount')
# plt.colorbar(label ='Customer Rating')
plt.ylabel('Distance Time (minutes)')
plt.xlabel('Distance (km)')
plt.show()
```

```python
plt.scatter(food['distance_km'],food['delivery_time_min'],c= food['rating'],cmap='RdYlGn',alpha=0.6,edgecolors='black',s=60)
plt.title('Late order Get Bad Rating')
plt.colorbar(label ='Customer Rating')
plt.ylabel('Distance Time (minutes)')
plt.xlabel('Distance (km)')
plt.show()
```

```python
veg = food[food['food_type'] == 'Veg']
nonveg = food[food['food_type'] =='Non-Veg']
plt.scatter(veg['distance_km'],veg['delivery_time_min'],color = '#2F9E44',alpha=0.6,label='Veg')
plt.scatter(nonveg['distance_km'],nonveg['delivery_time_min'],color = '#E03131',alpha=0.6,label='NonVeg')
plt.title('Veg Vs Nonveg orders')
plt.xlabel('Distance_km')
plt.ylabel(('Delivery Time in minuts'))
plt.legend()
plt.show()
```

# 3) Bar Chart : plt.bar()

```python
poll = pd.read_csv('street_food_poll.csv')
poll
```

```python
poll['total_votes'] = poll['boys_votes'] + poll['girls_votes']
poll
```

```python
plt.bar(poll['food'],poll['total_votes'])
plt.show()
```

```python
plt.figure(figsize=(9,4.5))
plt.bar(poll['food'],poll['total_votes'],color = '#F59F00',edgecolor = 'black',width=0.6)
plt.show()
```

```python
plt.figure(figsize=(9,4.5))
plt.bar(poll['food'],poll['total_votes'],color = '#F59F00',edgecolor = 'black',width=0.6)
plt.title('Class Poll -- Favourite Street Food')
plt.xlabel('Street food')
plt.ylabel('Total Votes')
plt.xticks(rotation = 20)
plt.show()
```

```python
plt.figure(figsize=(9,4.5))
bars = plt.bar(poll['food'],poll['total_votes'],color = '#F59F00',edgecolor = 'black',width=0.6)
plt.bar_label(bars,padding=7)
plt.title('Class Poll -- Favourite Street Food')
plt.xlabel('Street food')
plt.ylabel('Total Votes')
plt.xticks(rotation = 20)
plt.ylim(0,70)
plt.show()
```

```python
ranked = poll.sort_values('total_votes',ascending=False)
colors = ['#E03131'] +['#ADB5BD'] * 6
plt.figure(figsize=(9,4.5))
bars = plt.bar(ranked['food'],ranked['total_votes'],color = colors, edgecolor = 'black',width=0.6)
plt.bar_label(bars,padding=3)
plt.title('Pani Puri wins the class Poll')
plt.xlabel('Street Food')
plt.ylabel('Total Votes')
plt.xticks(rotation = 20)
plt.ylim(0,70)
plt.show()
```

# 4) Group & Stacked Bar Chart 

```python
x = np.arange(len(poll))
print(x)
```

```python
plt.bar(x-0.2,poll['boys_votes'],width=0.4,label='Boys')
plt.bar(x + 0.2, poll['girls_votes'],width =0.4,label = 'Girls')
plt.show()
```

```python
plt.figure(figsize = (9,4.5))
plt.bar(x-0.2,poll['boys_votes'],width=0.4,label='Boys',color ='Green')
plt.bar(x + 0.2, poll['girls_votes'],width =0.4,label = 'Girls',color='orange')
plt.xticks(x,poll['food'])
plt.title('Street Food Polls -- Boys Vs Girls')
plt.ylabel('Votes')
plt.legend(loc ='upper left')
plt.show()
```

```python
plt.figure(figsize = (9,4.5))
plt.bar(poll['food'],poll['boys_votes'],label='Boys',color ='Green')
plt.bar(poll['food'], poll['girls_votes'],label = 'Girls',color='orange',bottom = poll['boys_votes'])
# plt.xticks(x,poll['food'])
plt.title('Street Food Polls -- Stacked Votes')
plt.ylabel('Votes')
plt.legend()
plt.show()
```

```python
plt.figure(figsize = (9,4.5))
b1 = plt.bar(poll['food'],poll['boys_votes'],label='Boys',color ='Green')
b2 = plt.bar(poll['food'], poll['girls_votes'],label = 'Girls',color='orange',bottom = poll['boys_votes'])
plt.bar_label(b1,label_type='center',color = 'black')
plt.bar_label(b2,label_type='center',color = 'black')
totals =(poll['boys_votes'] + poll['girls_votes']) 
plt.bar_label(b2,labels =totals,padding=3)
# plt.xticks(x,poll['food'])
plt.title('Street Food Polls -- Stacked Votes')
plt.ylabel('Votes')
plt.legend()
plt.ylim(0,70)
plt.show()
```

```python
categories = ['rent','groceries','electricity','travel','entertainment','shopping']
yearly = expenses[categories].sum()
names = ['Rent','Groceries','electricity','Travel','Entertaintment','Shopping']
colours = ["#4C6EF5","#40C057","#FAB005","#15AABF","#BE4BDB","#FA5252"]
total = yearly.sum()
plt.pie(yearly,labels=names,autopct='%1.1f%%',colors=colours,startangle=90,pctdistance=0.78,
        wedgeprops=dict(width=0.45,edgecolor='white'))
plt.text(0,0,'Total\nRs'+str(total),ha='center',va='center',fontsize=12,fontweight='bold')
plt.title('Yearly Spending -- Dount Chart')
plt.show()
```

# 6) Box plot : plt.boxplot()

```python
marks = pd.read_csv('student_exam_marks.csv')
marks
```

```python
plt.boxplot(marks['maths'])
plt.show()
```

```python
q1 = marks['maths'].quantile(0.25)
median = marks['maths'].median()
q3 = marks ['maths'].quantile(.75)
top = marks['maths'].max()
plt.boxplot(marks['maths'])
plt.text(1.1, q1, 'Q1 (25%)', va='center')
plt.text(1.1,median,'Median (50%)',va = 'center')
plt.text(1.1,q3,'Q3(75%)', va ='center')
plt.text(1.05,top,'Upper Whisker', va = 'center')
# plt.text(1.05, outlier_value, 'Outlier', va='center', color='red')
plt.title('Math marks -- Parts of a box Plot')
plt.show()
```

```python
marks.describe()
```

```python
subjects = ['maths','science','english','computer']
names = ['Maths','Science','English','Computer']
plt.boxplot(marks[subjects],tick_labels=names)
plt.ylabel('marks')
plt.show()
```

```python
subjects = ['maths','science','english','computer']
names = ['Maths','Science','English','Computer']
colours = ['green','red','yellow','blue']
box = plt.boxplot(marks[subjects],tick_labels=names,patch_artist=True)
for patch ,color in zip (box['boxes'],colours):
    patch.set_facecolor(color)
plt.ylabel('marks')
plt.show()
```
