# 14) Handling Missing Values :

```python
import numpy as np
```

```python
import pandas as pd
```

```python
temp = pd.Series([72,75,np.nan,69,np.nan,80])
print(temp)
```

```python
missing_mask = temp.isnull()
print(missing_mask)
```

```python
missing_count = missing_mask.sum()
print(missing_count)
```

```python
dropped = temp.dropna()
print(dropped)
```

```python
average = temp.mean()
print(average)
```

```python
filled = temp.fillna(average)
print(filled)
```
