# 36) Route report between two area (function)

## all rides between two areas ,in both direction

```python
def route_report(area1, area2):
    a_to_b = (rides["pickup_area"] == area1) & (rides["drop_area"] == area2)
    b_to_a = (rides["pickup_area"] == area2) & (rides["drop_area"] == area1)
    trips = rides[a_to_b | b_to_a]
    return trips[["pickup_area", "drop_area", "ride_type", "fare"]]
```
