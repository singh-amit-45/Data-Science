## Filtering  a DataFrame 

# 28) Find All Airport Ribes :

```python
rides.head(2)
```

```python
to_airport = rides['drop_area'].str.contains('Airport')
to_airport
```

```python
from_airport = rides['drop_area'].str.contains('Airport')
from_airport
```

```python
airport_rides = rides[to_airport | from_airport]
airport_rides[['city','pickup_area','drop_area','distance_km']].head()
```

# 29) How many rides had surge pricing :

```python
surge_rides = rides[rides['surge']=="Y"]
surge_rides[['city','hour','weather','fare']].head()
```

# 30) Rides of one type cancelled in one City:

```python
in_pune = rides['city'] =='Pune'
is_auto = rides['ride_type'] =='Auto'
cancelled = rides['status']=='Cancelled'
```

```python
pune_auto_cancelled = rides[in_pune & is_auto & cancelled]
pune_auto_cancelled
```

```python
print(pune_auto_cancelled[['date','pickup_area','drop_area','cancelled_by']])
```

```python
pune_auto_cancelled[['date','pickup_area','drop_area','cancelled_by']].shape[0]
```
