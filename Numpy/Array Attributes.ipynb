# Array Attribute
## a.ndim , a.shape , a.size , a.iteamsize , a.dtype

```python
# !pip install numpy
```

```python
import numpy as np
```

```python
a1 = np.arange(10,dtype=np.int32)
a2 = np.arange(12,dtype=float).reshape(3,4)     # reshape means row and col
a3 = np.arange(8).reshape(2,2,2)

print("a1:",a1)
print("a2:",a2)
print("a3:",a3)
```

```python
print(a1.ndim)
print(a2.ndim)  # .ndim means the number of dimensions of an array.1D,,2D,3D
print(a3.ndim)
```

```python
print(a1,a1.shape) # .shape tells you the size of each dimension.
print(a2,a2.shape)
print(a3,a3.shape)
```

```python
print(a1,a1.size)  # .size tells you the total number of values in the array or .size means array mein total number of elements.
print(a2,a2.size)
print(a3,a3.size)
```

```python
print(a1,a1.dtype)     #.dtype tells you data type.
print(a2,a2.dtype)
print(a3,a3.dtype)
```

# Changing data types.

```python
prices = np.array([10.9,20.2,30.8])
print(prices)
whole = prices.astype(int)
print(whole)
```

```python
back_to_float = whole.astype(np.float64) # .astype is used for changing the data type.
print(back_to_float)
```

```python
ids = np.array([101,102,103])
print(ids)
as_text = ids.astype(str)
print(as_text)
```

```python
flags = np.array ([0,4,0,5,1,0])
print(flags)
as_bool = flags.astype(bool)
print(as_bool)
```
