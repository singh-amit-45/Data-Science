```python
vals = np.array([1,2,3,4,7,8])
print(vals)
```

```python
print(np.round(np.log(vals),2)) #the natural logarithm (logₑ) of every value in vals and
# 2 means round 2 decimal place
```

```python
print(np.round(np.sin(vals),2)) # calculates the sine of each value in vals, rounds each answer to 2 decimal places, and prints it.
```

```python
print(np.floor(vals)) # np.floor(vals) rounds every value in vals downward to the nearest whole number.
print(np.ceil(vals))  # np.ceil(vals) means: round every value in vals upward to the nearest whole number.
```

```python
m =np.array([[1,2],[3,4]])
print(m)
r = m.ravel()
print(r)
```

```python
r[0] = 99
print(m)
print(r)
```

```python
s = m.flatten()  # .flatten() means: make an array flat—convert 2D or 3D data into one single row.
print(s)
```

### ravel() usually returns a view of the original array. Changing it can change the original array.
### flatten() always returns a copy. Changing it does not change the original array.

```python
s = m.flatten()
print(s)
```

```python
s[2]=555
print(m)
print(s)
```

# Stacking

```python
a4 = np.arange(12).reshape(3,4)
print(a4)
a5 = np.arange(12,24).reshape(3,4)
print(a5)
```

```python
print(np.hstack((a4,a5))) # .hstack() joins arrays horizontally
print(np.vstack((a4,a5))) # .vstack() joins arrays vertically
```

```python
b1 = np.array([1,2,3])
b2 = np.array([4,5,6])
print(b1)
print(b2)
print(np.stack((b1,b2),axis=0)) # row wise
print(np.stack((b1,b2),axis=1)) # colum wise
```

```python
print(a4)
print(a5)
```

```python
print(np.concatenate((a4,a5),axis=0)) 
print(np.concatenate((a4,a5),axis=1))
```

# Spliting

```python
a4 = np.arange(20).reshape(4,5)
print(a4)
```

```python
parts = np.hsplit(a4,5)
print(parts)
```

```python
a5 = np.arange(12,24).reshape(3,4)
print(a5)
```

```python
top,bottom = np.vsplit(a5,[2])
print(top)
print(bottom)
```

```python
a3 = np.arange(24).reshape(2,3,4)
print(a3)
```

```python
left,right = np.split(a3,2,axis=2)
print(left)
print(right)
```
